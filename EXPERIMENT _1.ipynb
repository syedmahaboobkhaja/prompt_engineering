```python

import os
from getpass import getpass
from groq import Groq

# ==========================================
# 1. Get Groq API Key
# ==========================================
api_key = getpass("Enter your GROQ API key: ")

if not api_key:
    print("❌ Error: GROQ_API_KEY was not provided.")
else:
    os.environ["GROQ_API_KEY"] = api_key

# ==========================================
# 2. Check API Key
# ==========================================
try:
    if not os.getenv("GROQ_API_KEY"):
        raise ValueError("GROQ_API_KEY not found.")

    print("✅ Environment ready.")
    print("✅ Groq API key loaded successfully.")

    # ==========================================
    # 3. Create Groq Client
    # ==========================================
    client = Groq(api_key=os.getenv("GROQ_API_KEY"))

    # ==========================================
    # 4. Define Prompt
    # ==========================================
    prompt = "What is prompt engineering? Answer in one sentence."

    print("\nPrompt:")
    print(prompt)

    # ==========================================
    # 5. Send Prompt to Groq
    # ==========================================
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    # ==========================================
    # 6. Get Model Response
    # ==========================================
    answer = response.choices[0].message.content

    print("\nResponse:")
    print(answer)

    # ==========================================
    # 7. Optional Connectivity Test
    # ==========================================
    second_prompt = "Explain artificial intelligence in one sentence."

    response2 = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": second_prompt
            }
        ]
    )

    print("\n--- Connectivity Test ---")
    print("Second Prompt:")
    print(second_prompt)

    print("\nSecond Response:")
    print(response2.choices[0].message.content)

    print("\n✅ Groq connectivity test completed successfully.")

# ==========================================
# 8. Error Handling
# ==========================================
except ValueError as e:
    print(f"\n❌ Configuration Error: {e}")

except Exception as e:
    print(f"\n❌ API/Network Error: {e}")
```
