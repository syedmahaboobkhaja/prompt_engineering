```python
# ============================================================
# EXPERIMENT 2: PROMPT ENGINEERING
# Baseline Prompt vs Enhanced Prompt
# Groq API + Google Colab
# ============================================================

import os
from getpass import getpass
from groq import Groq

# ------------------------------------------------------------
# STEP 1: Load API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Define send_prompt() Function
# ------------------------------------------------------------

def send_prompt(prompt):

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.7,
        max_tokens=200
    )

    return response.choices[0].message.content


# ------------------------------------------------------------
# STEP 4: Define Baseline Prompt
# ------------------------------------------------------------

baseline_prompt = "Explain what is recursion."


# ------------------------------------------------------------
# STEP 5: Define Enhanced Prompt
# ------------------------------------------------------------

enhanced_prompt = """
You are a computer science professor teaching beginners.
Explain recursion using a simple real-world analogy.
Give a clear definition, explain the base case, and provide
one simple Python example. Keep the explanation under 100 words.
"""


# ------------------------------------------------------------
# STEP 6: Send Baseline Prompt
# ------------------------------------------------------------

try:

    baseline_response = send_prompt(
        baseline_prompt
    )

except Exception as e:

    baseline_response = f"Error: {e}"


# ------------------------------------------------------------
# STEP 7: Send Enhanced Prompt
# ------------------------------------------------------------

try:

    enhanced_response = send_prompt(
        enhanced_prompt
    )

except Exception as e:

    enhanced_response = f"Error: {e}"


# ------------------------------------------------------------
# STEP 8: Display Baseline Result
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("BASELINE PROMPT")
print("=" * 70)

print("\nPrompt:")
print(baseline_prompt)

print("\nResponse:")
print(baseline_response)


# ------------------------------------------------------------
# STEP 9: Display Enhanced Result
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("ENHANCED PROMPT")
print("=" * 70)

print("\nPrompt:")
print(enhanced_prompt)

print("\nResponse:")
print(enhanced_response)


# ------------------------------------------------------------
# STEP 10: Simple Comparison
# ------------------------------------------------------------

baseline_words = len(baseline_response.split())
enhanced_words = len(enhanced_response.split())

print("\n")
print("=" * 70)
print("COMPARISON")
print("=" * 70)

print(f"\nBaseline response length : {baseline_words} words")
print(f"Enhanced response length : {enhanced_words} words")

print("\nAnalysis:")
print("• Baseline prompt gives a general explanation.")
print("• Enhanced prompt provides role and context.")
print("• Enhanced prompt specifies the required content.")
print("• Enhanced prompt provides a length constraint.")
print("• Enhanced prompting generally produces more structured output.")
```
